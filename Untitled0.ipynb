### Optimized YOLOv11 Training Configuration
This configuration applies recommended hyperparameters for the Tesla T4 to achieve highly stable and accurate training.

```python
from ultralytics import YOLO

# Best hyperparameter settings for custom YOLOv11 target detection
model = YOLO("yolo11n.pt")

model.train(
    data="/content/My-First-Project-1/data.yaml",
    epochs=100,             # Stable epoch size
    imgsz=640,              # Target image resolution
    batch=16,               # Explicit batch size to prevent dataloader thread bottlenecks on T4
    device=0,               # Use GPU
    workers=2,              # Safe worker count to prevent CPU/RAM memory leaks in Colab
    optimizer="AdamW",      # Best general-purpose optimizer for convergence speed
    lr0=0.001,              # Balanced initial learning rate
    lrf=0.01,               # Final learning rate fraction
    momentum=0.937,         # Standard SGD/Adam momentum
    weight_decay=0.0005,    # Regularization to prevent overfitting
    warmup_epochs=3.0,      # Give the learning rate time to ramp up safely
    close_mosaic=10,        # Turn off mosaic augmentation in the last 10 epochs for precision
    cos_lr=True,            # Cosine learning rate scheduler
    patience=20,            # Early stopping patience limit
    project="/content/drive/MyDrive/yolo-runs",
    name="optimized-yolo",
    exist_ok=True,
)
```

### Detailed Model Evaluation (Industry Metrics)
In object detection, instead of standard regression $R^2$ scores, we evaluate performance using Localization and Classification metrics:
- **mAP50 & mAP50-95**: Mean Average Precision (overall model accuracy)
- **Precision & Recall**: Classification correctness and detection coverage
- **F1-Score**: Harmonic mean of Precision and Recall
- **Confusion Matrix**: Class-to-class misclassification visualization

### Deploying YOLOv11 with FastAPI
Below is the complete, self-contained implementation of a FastAPI application. This application reads an uploaded image file, processes it through your trained `best.pt` model, and returns standard JSON metadata detailing detected classes, bounding boxes, and confidence scores. This script can be saved locally as `app.py` and run on any server with PyTorch and FastAPI installed.

```python
fastapi_code = """from fastapi import FastAPI, File, UploadFile, HTTPException
from fastapi.responses import JSONResponse
from ultralytics import YOLO
from PIL import Image
import io
import uvicorn

# Initialize FastAPI app
app = FastAPI(
    title="YOLOv11 Custom Detection API",
    description="Microservice to serve our custom YOLOv11 target detection model.",
    version="1.0.0"
)

# Path to your trained model weights
MODEL_PATH = "/content/drive/MyDrive/yolo-runs/optimized-yolo/weights/best.pt"

try:
    # Load the YOLO model once during startup
    model = YOLO(MODEL_PATH)
    print(f"Successfully loaded YOLOv11 model from {MODEL_PATH}")
except Exception as e:
    print(f"Error loading model at startup: {e}")
    model = None

@app.get("/")
def read_root():
    return {"status": "online", "model_loaded": model is not None}

@app.post("/predict")
async def predict(file: UploadFile = File(...)):
    if not model:
        raise HTTPException(status_code=500, detail="Model weights are not loaded on server.")

    # Validate upload mime-type
    if file.content_type not in ["image/jpeg", "image/png", "image/jpg"]:
        raise HTTPException(status_code=400, detail="Invalid image file type. Upload JPG or PNG.")

    try:
        # Read file into memory and convert to PIL Image
        image_data = await file.read()
        image = Image.open(io.BytesIO(image_data)).convert("RGB")

        # Perform Inference
        results = model.predict(source=image, conf=0.25)
        result = results[0] # Single image prediction

        # Construct JSON output response
        predictions = []
        for box in result.boxes:
            # Get coordinates in standard format: [x_min, y_min, x_max, y_max]
            coords = box.xyxy[0].tolist()
            class_id = int(box.cls[0].item())
            class_name = result.names[class_id]
            confidence = float(box.conf[0].item())

            predictions.append({
                "class_id": class_id,
                "class_name": class_name,
                "confidence": round(confidence, 4),
                "box": {
                    "x_min": round(coords[0], 2),
                    "y_min": round(coords[1], 2),
                    "x_max": round(coords[2], 2),
                    "y_max": round(coords[3], 2)
                }
            })

        return JSONResponse(content={
            "filename": file.filename,
            "num_detections": len(predictions),
            "predictions": predictions
        })

    except Exception as e:
        raise HTTPException(status_code=500, detail=f"Failed to process image: {str(e)}")

if __name__ == '__main__':
    # Host on all interfaces on port 8000
    uvicorn.run(app, host="0.0.0.0", port=8000)
"""

# Save code block locally to a file so user can download it
with open('app.py', 'w') as f:
    f.write(fastapi_code)

print("FastAPI app.py code successfully created and saved to the runtime memory!")
```
